```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Sample data
data = {
    "Hours_Studied": [1, 2, 3, 4, 5, 6],
    "Exam_Score": [35, 40, 50, 55, 65, 70]
}

df = pd.DataFrame(data)

# Display the data
print(df)
```

```python
# Prepare the input and output variables
X = df[["Hours_Studied"]]
y = df["Exam_Score"]

# Create the model
model = LinearRegression()

# Train the model
model.fit(X, y)

print("Linear Regression model trained successfully!")
print("Model coefficient:", model.coef_[0])
print("Model intercept:", model.intercept_)
```

```python
# Predict the score for 7 hours of study
new_student = pd.DataFrame({"Hours_Studied": [7]})

predicted_score = model.predict(new_student)

print("Hours studied:", 7)
print("Predicted exam score:", round(predicted_score[0], 2))
```

```python
# Plot the original data
plt.scatter(X["Hours_Studied"], y, label="Actual scores")

# Draw the regression line
plt.plot(
    X["Hours_Studied"],
    model.predict(X),
    label="Regression line"
)

# Add labels and title
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.title("Linear Regression: Study Hours vs Exam Score")
plt.legend()
plt.grid(True)

# Display the graph
plt.show()
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Predict scores for the original data
y_pred = model.predict(X)

# Evaluate the model
mae = mean_absolute_error(y, y_pred)
mse = mean_squared_error(y, y_pred)
r2 = r2_score(y, y_pred)

print("Mean Absolute Error (MAE):", round(mae, 2))
print("Mean Squared Error (MSE):", round(mse, 2))
print("R-squared (R²):", round(r2, 3))
```
