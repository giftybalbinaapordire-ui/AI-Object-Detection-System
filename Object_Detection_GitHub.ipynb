```python
from pathlib import Path
import shutil

project = Path.cwd().parent
dataset = project / "dataset"

train_images = dataset / "images" / "train"
train_labels = dataset / "labels" / "train"
val_images = dataset / "images" / "val"
val_labels = dataset / "labels" / "val"

# Get the image files currently in the training folder
images = sorted([
    f for f in train_images.iterdir()
    if f.suffix.lower() in [".jpg", ".jpeg", ".png"]
])

# Use the last image for validation
if len(images) >= 2:
    image = images[-1]
    label = train_labels / (image.stem + ".txt")

    if label.exists():
        shutil.move(str(image), str(val_images / image.name))
        shutil.move(str(label), str(val_labels / label.name))
        print("Validation image:", image.name)
        print("Validation label:", label.name)
        print("Remaining training images:", len(images) - 1)
    else:
        print("Matching label not found:", label.name)
else:
    print("Not enough training images to create a validation set.")
```

```python
from pathlib import Path
import shutil

project = Path.cwd().parent
dataset = project / "dataset"

raw_images = dataset / "raw_images"
annotations = dataset / "annotated_images"

train_images = dataset / "images" / "train"
train_labels = dataset / "labels" / "train"

copied = 0

for image in raw_images.iterdir():
    if image.suffix.lower() not in [".jpg", ".jpeg", ".png"]:
        continue

    label = annotations / (image.stem + ".txt")

    if label.exists():
        shutil.copy2(image, train_images / image.name)
        shutil.copy2(label, train_labels / label.name)
        print("Copied:", image.name)
        copied += 1

print("\nTotal matched image-label pairs copied:", copied)
```
